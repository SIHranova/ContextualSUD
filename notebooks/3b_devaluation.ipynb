## Gradual outcome devaluation and habit expression

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
import scipy.special as scp
import matplotlib.ticker as ticker
# import jsonpickle
from itertools import product
from matplotlib.ticker import MultipleLocator
from scipy.optimize import curve_fit

from utilities import run_simulation, create_long_df_gradual_change, plot_reward_regime, fit_posterior_traces
from classes import *

sns.set_style("whitegrid")
dpi = 200

#############################################
# setup simulation run, agent, env parameters
############################################

# originally no noise in reward init counts
np.random.seed(1)
pars = {}
pars["hs"] = np.array([1000,10,1])
h_order = ["no habit", "moderate_habit", "strong habit"]
h_order = {h:h_order[hi] for hi, h in enumerate(pars["hs"])}
pars["k_range"] = [0.02, 3.]
env_order = ["low certainty", "high certainty"]
env_order = {k:env_order[ki] for ki, k in enumerate(pars["k_range"])}
pars["context_volatility"] = 0.99
pars["changing_rewards"] = False
pars["changing_utility"] = True
pars["nsims"]= 200
pars["debug"] = False

# agent and env parameters
pars["TAU"] = 400
pars["T"] = 2
pars["na"] = 2
pars["ns"] = pars["na"] + 1
pars["no"] = pars["ns"]
pars["nr"] = 3
pars["nc"] = 2
pars["npi"] = pars["na"]**(pars["T"]-1)
pars["train_trials"] = 0
pars["switch"] = pars["TAU"]//2
pars["training_protocol"] = [np.array([0]*(pars["switch"]+pars["train_trials"]) + [1]*pars["switch"])]

# create utility decaying with rates k in k_range
start = 1
stop = 0
logistic = lambda x, k: stop - (start-stop) / (np.exp(pars["switch"]*k) - 1) \
                        + (np.exp(pars["switch"]*k) + 1) / (np.exp(pars["switch"]*k) - 1)\
                        * (start-stop)/(1+np.exp(k*(x-pars["switch"])))
utilities = []
for k in pars["k_range"]:
    x = np.arange(0,pars["switch"]*2,1)
    p_win = logistic(x,k)
    utility = np.array([start+stop-p_win, p_win, np.array([1-(start+stop)]*pars["TAU"])]).T
    assert(np.all(utility.sum(axis=1)) == 1)
    utilities.append([utility,k])
plot_reward_regime(pars["k_range"], pars["switch"], pars["TAU"], start, stop, dpi=dpi)


# create env reward generation for different contexts; rewards do not change between contexts
p_win = 0.9
Rho = np.zeros([pars["nr"],pars["ns"],pars["nc"]])
Rho[:,:,0]= np.array([[1-p_win, p_win  , 0],\
                      [p_win  , 1-p_win, 0],\
                      [0      , 0      , 1]])
Rho[:,:,1]= np.array([[1-p_win, p_win  , 0],\
                      [p_win  , 1-p_win, 0],\
                      [0      , 0      , 1]])
pars["Rho"] = Rho

################################################
# create pars dictionary for each simulation run
################################################

agent_instance = np.arange(pars["nsims"])
context_volatilities = np.array([pars["context_volatility"]])
parameter_combinations = product(agent_instance, pars["hs"], pars["training_protocol"], context_volatilities, utilities)

sim_params = []
for pi, p in enumerate(parameter_combinations):
    par = pars.copy()
    par["rep"] = p[0]
    par["h"] = p[1]
    par["training_protocol"] = p[2]
    par["context_volatility"] = p[3]
    par["utility"] = p[4][0]
    par["decay_rate"] = p[4][1]
    par["sim"] = pi
    sim_params.append(par)

##################################
# RUN ALL SIMULATIONS
##################################
long_dfs = []
for pi, par in enumerate(sim_params):
    print(f"just finished sim {par["sim"]}") if pi% 250 == 0 else None
    world, _ , _ = run_simulation(**par)
    par["train_trials"] = par.pop("train_trials")
    long_dfs.append(create_long_df_gradual_change(world, par, env_order, h_order))

env_order = [item for key, item in env_order.items()] 
h_order = [item for key, item in h_order.items()]
long_df = pd.concat(long_dfs)
long_df = long_df.astype({"action":"category", "habitual_tendency": "category", "utility_decay_rate": "category", 
                          "h": pd.CategoricalDtype(categories=h_order, ordered=True),
                          "environment_type": pd.CategoricalDtype(categories=env_order, ordered=True)})
name = f"data_devaluation_{pars["TAU"]}_util{start}_context{pars["context_volatility"]}"
long_df.to_csv(name + "_long.csv", index=False)
```
