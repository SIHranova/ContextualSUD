## Motivation for a modelling paper: Context Inference and Habitual Action in SUD

Substance use disorder (SUD) is characterized by persistent drug-seeking actions that are often described as habitual: 
- behavior becomes resistant to outcome devaluation and contingency degradation
- is rapidly reinstated by cues or contexts
- tends to relapse after extinction 

A large experimental literature, especially in fear conditioning and extinction, suggests that such relapse phenomena arise because animals infer latent contexts. In this view, extinction does not erase earlier learning but rather creates a new context-specific memory; returning to the original context reactivates the earlier memory and behavior.
Several computational models have formalized this idea using latent context inference, successfully explaining phenomena such as 
- reacquisition (CS+,CS0, CS+ or instrumentally changes in contingencies +,-,+, also known as savings) 
- renewal - ABA, ABC etc
- reinstatement (CS+,CS0,+,CS0 - expectation of CS+)
- spontaneous recovery (the spontaneous recovery of behaviour with the passage of time) Gershman only, not sure whether anyone else has used context inference to model these effects). However, most of these models (from Gershman, Niv) describe behavior primarily through outcome prediction. By contrast, SUD involves instrumental action policies, i.e., drug-seeking actions that can become highly automatized through repetition. In this domain, the role of habitual action control has long been emphasized.

We propose that a mechanistic account of SUD-like behavior requires combining two ingredients:
1.	Context inference: agents maintain uncertainty about which latent context is currently active.
2.	Context-specific action priors: repeated behavior builds strong priors over actions within each context. (Schwoebel et al 2021)

In this framework, behavior depends jointly on the inferred context and the strength of action priors. Importantly, this allows habitual responding to emerge not only from extensive training but also from the interaction between context uncertainty and strong policy priors. Conceptually, this provides two independent mechanismsshaping habit-like behavior:
- Context bias: a tendency to infer certain contexts (e.g., drug-associated contexts) with high probability. For example, we might assume that vulnerable people have a lower inferred context uncertainty in general.
- Habit strength: the precision of action priors within a context.

## Basic pavlovian cue simulation

Following graph represents a pavlovian trial, where S is the CS, r is the US, c is context and theta learned parameters of association between CS and US. Resulting generative model is 

$$p(\theta)p(\lambda)p(o|c,\lambda)p(c)p(s)p(r|s,c,\theta)$$

Below are implemented the mean field updates for approximate posterior $q(c)$, $q(\lambda)$ and $q(\theta)$

Belief update in different pavlovian conditioning tasks can be simulated by varying CS,US and cue combinations.

```python
from matplotlib import pyplot as plt
import networkx as nx
import numpy as np



np.random.seed(6)
graph = nx.DiGraph()
graph.add_edges_from([("s","r"),("c","r"),(r"$\theta$", "r"), ("c","o"), (r"$\lambda$", "o")])

# position = nx.spring_layout(graph)
nx.draw_networkx(graph)#, position)

```

```python
### initial values
import seaborn as sns
import scipy.special as scp

nt = 30
cs = [1]*nt + [1]*nt + [1]*nt
us = [1]*nt + [1]*nt + [0]*nt
context_obs = [0]*nt + [1]*nt + [1]*nt
T = len(cs)

np.random.seed(1)

nc = 3   # number of contexts
no = 3   # number of contextual cues
ns = 2   # number of CS (light on or off)
nr = 2   # number of rewards/punishments; simulating fear conditioning so 0=no punishment and 1=punishment
npi = 2  # number of policies; 0=no fear response, 1=fear response

theta = np.ones([nr,ns,nc]) + np.random.uniform(size=[nr,ns,nc])*0.2
p_rewards = theta / theta.sum(axis=0)

lambdah = np.ones([no,nc]) + np.random.uniform(size=[no,nc])*0.2
p_context_obs = lambdah/lambdah.sum(axis=0)

alpha = np.ones([T,npi,nc])

actions = np.zeros(T-1)

q = 0.95
trans_matrix = np.eye(nc)*0.95
trans_matrix[trans_matrix != q] = (1-q)/(nc-1)
assert(np.all(trans_matrix.sum(axis=0)==1))

p_c = np.array([0.95]+[1-0.95]*(nc-1))
q_c = np.zeros([T,nc])
q_c[0] = p_c

q_policy_given_context = np.zeros([T,npi,nc])

for t in range(1,T):

    o = np.eye(no,dtype=int)[context_obs[t]]
    r = np.eye(nr,dtype=int)[us[t]]
    s = np.eye(ns,dtype=int)[cs[t]]

    # infer posterior over contexts
    q_c[t] = np.log(p_c) + np.einsum("r,s,rsc->c",r,s,np.log(p_rewards)) + np.dot(np.log(p_context_obs).T,o)
    q_c[t] = np.exp(q_c[t])/ np.exp(q_c[t]).sum()
    
    # updates CS-US pairing strength
    theta[np.argmax(r),np.argmax(s),:] += q_c[t]
    p_rewards = scp.softmax(scp.digamma(theta) - scp.digamma(theta.sum(axis=0)),axis=0)           # should be digamma
    
    # update observation x context mapping
    lambdah[np.argmax(o),:] += q_c[t]             # should be digamma
    p_context_obs = scp.softmax(scp.digamma(lambdah)-scp.digamma(lambdah.sum(axis=0)))

    action = np.argmax(q_policy_given_context[t].dot(q_c[t]))
    actions[t-1] = action

    alpha[t] = alpha[t-1].copy()
    alpha[t,action] = alpha[t,action] + q_c[t]
    q_policy_given_context[t] = scp.softmax(scp.digamma(alpha[t]) - scp.digamma(alpha[t].sum(axis=0)),axis=0)
    # learn context-specific policy prior
    p_c = trans_matrix.dot(q_c[t])


####
fig, axes = plt.subplots(1,nc,figsize=(10,3))
fig.set_tight_layout = True

for ai, ax in enumerate(axes):
    sns.heatmap(theta[:,:,ai],ax=ax,annot=True)
    ax.set_title(rf"$p(r|s,c={ai})$")

####
fig, ax = plt.subplots(1,1)
fig.set_tight_layout = True

sns.heatmap(lambdah[:,:],ax=ax,annot=True)
ax.set_title(r"$p(o|c)$")


####
fig, ax = plt.subplots(1,1)
sns.heatmap(q_c.T,ax=ax)
ax.set_xlabel("t")
ax.set_ylabel(r"$q(c)$")

```

```python
### initial values
import seaborn as sns
import scipy.special as scp
import pandas as pd


nt = 30

# row 1 - context obs; row 2 - CS; row 3 - US
ABA = np.array([["A","B","A"],[1,1,1],[1,0,0]],dtype=object)
ABB = np.array([["A","B","B"],[1,1,1],[1,0,0]],dtype=object)

T = nt*3

experiments = {"ABA":ABA, "ABB":ABB}

n_sims = 100
np.random.seed(1)

dfs = []

for key in experiments:
    for habitual_tendency in [1,1000]:
        for sim in range(n_sims):

            nc = 2   # number of contexts
            no = 2   # number of contextual cues
            ns = 2   # number of CS (light on or off)
            nr = 2   # number of rewards/punishments; simulating fear conditioning so 0=no punishment and 1=punishment
            npi = 2  # number of policies; 0=no fear response, 1=fear response

            experiment = experiments[key].copy()
            context_obs_letters = experiment[0]
            context_obs_mapping = {"A":0, "B":1, "C":2}
            context_obs = np.array([context_obs_mapping[c] for c in context_obs_letters])
            experiment[0] = context_obs
            context_contingencies, true_context = np.unique(experiment.astype(int),axis=1, return_inverse=True)

            
            context_obs = context_obs.repeat(nt)
            cs = experiment[1].repeat(nt) # [1]*nt + [1]*nt + [1]*nt  # CS sequence for training stages
            us = experiment[2].repeat(nt) #[1]*nt + [1]*nt + [0]*nt  # outcome sequence (punishment or safe)
            true_context = context_obs.copy()   # here it will depend on the combination of context_obs and us
            
            exp_phase = np.array(["training", "extinction", "test"]).repeat(nt)

            # associative parameters for CS->US mapping
            beta = np.ones([nr,ns,nc]) + np.random.uniform(size=[nr,ns,nc])*0.2
            p_rewards = beta / beta.sum(axis=0)

            # context-observation mapping parameters
            epsilon = np.ones([no,nc]) + np.random.uniform(size=[no,nc])*0.2
            p_context_obs = epsilon/epsilon.sum(axis=0)

            # policy prior counts (habit strength per context)
            alpha = np.ones([T,npi,nc])

            actions = np.zeros(T)

            # context transition prior (sticky contexts)
            q = 0.95
            trans_matrix = np.eye(nc)*0.95
            trans_matrix[trans_matrix != q] = (1-q)/(nc-1)
            assert(np.all(trans_matrix.sum(axis=0)==1))

            # initial latent context belief
            p_c = np.array([0.95]+[1-0.95]*(nc-1))
            q_c = np.zeros([T,nc])
            q_c[0] = p_c

            # policy arrays
            q_pi = np.zeros([T,npi])

            alpha = np.zeros([T,npi,nc])
            alpha[0] = np.ones([npi,nc])*habitual_tendency

            p_policy_context = np.zeros([T,npi,nc])
            p_policy_context[0] = alpha[0] / alpha[0].sum(axis=0)

            state_transition = np.stack([[[0.5,0],[0.5,1]],[[0.5,1],[0.5,0]]],axis=-1)  # implements logic of p(danger_{t}|danger_{t-1},fear response)
            state_transition = np.stack([[[1,0],[0,1]],[[0,1],[1,0]]],axis=-1)  # implements logic of p(danger_{t}|danger_{t-1},fear response)

            utility = np.array([0.8,0.02])
            state_utility = np.einsum('pra,p->ra',state_transition, utility)

            for t in range(1, T):
                a = np.random.rand()
                if a > 0.9:   
                    us[t] = int(not us[t])
                o = np.eye(no, dtype=int)[context_obs[t]]
                r = np.eye(nr, dtype=int)[us[t]]
                s = np.eye(ns, dtype=int)[cs[t]]

                # infer context posterior from prior, sensory and reward observations
                q_c[t] = np.log(p_c) + np.einsum("r,s,rsc->c", r, s, np.log(p_rewards)) + np.dot(np.log(p_context_obs).T, o)
                q_c[t] = np.exp(q_c[t]) / np.exp(q_c[t]).sum()

                # evaluate policies and combine with context-specific policy priors
                F_pi_c = np.log(state_utility).T.dot(r)
                q_pi_c = (p_policy_context[t-1] * scp.softmax(F_pi_c)[:, None]) / (p_policy_context[t-1] * scp.softmax(F_pi_c)[:, None]).sum(axis=0)
                # print(f"==>> scp.softmax(F_pi_c): {scp.softmax(F_pi_c)}")
                q_pi[t] = q_pi_c.dot(q_c[t])

                # choose action by maximum a posteriori policy
                action = np.random.choice(np.arange(npi), p=q_pi[t])
                actions[t-1] = action

                # update US prediction counts by context posterior
                beta[np.argmax(r), np.argmax(s), :] += q_c[t]
                p_rewards = scp.softmax(scp.digamma(beta) - scp.digamma(beta.sum(axis=0)), axis=0)

                # update context-cue counts
                epsilon[np.argmax(o), :] += q_c[t]
                p_context_obs = scp.softmax(scp.digamma(epsilon) - scp.digamma(epsilon.sum(axis=0)), axis=0)

                # habit learning: update policy prior counts context-specifically
                alpha[t] = alpha[t-1].copy()
                alpha[t, action] += q_c[t]
                p_policy_context[t] = scp.softmax(scp.digamma(alpha[t]) - scp.digamma(alpha[t].sum(axis=0)), axis=0)

                # update prior over contexts for next trial
                p_c = trans_matrix.dot(q_c[t])

            # label switch
            best_fit = 0
            best_labels = None

            for it in range(100):
                labels = np.random.permutation(nc)
                fit = np.sum(labels[np.argmax(q_c, axis=1)] == true_context) / len(true_context)

                if fit > best_fit:
                    best_fit = fit
                    best_labels = labels.copy()


            q_c = q_c[:,best_labels]
            beta = beta[:,:,best_labels]

            p_rewards - p_rewards[:,:,best_labels]
            epsilon = epsilon[:,best_labels]
            p_context_obs = p_context_obs[:,best_labels]

            alpha = alpha[:,:,best_labels]
            p_policy_given_context = p_policy_context[:,:,best_labels]

            df = pd.DataFrame({})
            df["q_c"] = q_c.flatten()
            df["context"] = np.tile(np.arange(nc), T)
            df["true_context"] = np.repeat(true_context, nc)
            df["sim"] = sim
            df["t"] = np.arange(T).repeat(nc)
            df["h"] = habitual_tendency
            df["actions"] = actions.repeat(nc)
            df["q_pi1"] = q_pi[:,0].repeat(nc)
            df["q_pi2"] = q_pi[:,1].repeat(nc)
            df["counts_pi1"] = alpha[:,0,:].flatten()
            df["counts_pi2"] = alpha[:,1,:].flatten()
            df["experiment"] = key
            df["cs"] = cs.repeat(nc)
            df["us"] = us.repeat(nc)
            df["context_obs"] = context_obs_letters.repeat(nt).repeat(nc)
            df["phase"] = exp_phase.repeat(nc)
            df["phase"] = df["phase"].astype(pd.CategoricalDtype(categories=["training", "extinction", "test"], ordered=True))  
            df["p_context_obs"] = [p_context_obs]*(T*nc)
            df["p_rewards"] = [p_rewards]*(T*nc)
            dfs.append(df)

dfs = pd.concat(dfs, ignore_index=True)




#### TASK SETUP
fig, ax = plt.subplots(1,1,figsize=(10,3))
fig.set_tight_layout = True
sns.heatmap(np.stack([context_obs,cs,us]).astype(int),ax=ax, annot=True)
ax.set_title("Task Parameters")
ax.set_xlabel("trial")
ax.set_yticklabels(["Obs","CS","US"])

# #### CS-US association counts plot
# fig, axes = plt.subplots(1,nc,figsize=(10,3))
# fig.set_tight_layout = True

# for ai, ax in enumerate(axes):
#     sns.heatmap(beta[:,:,ai],ax=ax,annot=True)
#     ax.set_title(rf"$p(r|s,c={ai})$")


# #### Learned observations maping
# fig, ax = plt.subplots(1,1)
# fig.set_tight_layout = True
# sns.heatmap(epsilon,ax=ax,annot=True)
# ax.set_title(r"$p(o|c)$")


# #### Context posterior
# fig, ax = plt.subplots(1,1)
# sns.heatmap(q_c.T,ax=ax)
# ax.set_xlabel("t")
# ax.set_ylabel(r"$q(c)$")


# #### Policy Prior
# fig, ax = plt.subplots(1,1)
# fig.set_tight_layout = True

# sns.heatmap(alpha[t]/alpha[t].sum(axis=0),ax=ax,annot=True)
# ax.set_title(r"$p(\pi|c)$")


```
