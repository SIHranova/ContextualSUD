```python
import pandas as pd
import seaborn as sns
data = pd.read_csv("PIT_B03_data.csv")
```

```python
for col in data.columns:
    print(col, data[col].dtype)
```

```python
import matplotlib.pyplot as plt

cols = ["subj","trial","InstrStim", "InstrResp", "PavlovianStimulus", "Correct", "Responses", "InsResp", "Congruity", "joypeakv","joytotalT"]

df = data[cols]

for col in df.columns:
    print(col, df[col].unique()[:6])

plt.grid()
sns.barplot(data=df, x="Congruity",y="Correct",hue="InstrStim")

plt.figure()
plt.grid()
sns.barplot(data=df, x="Congruity",y="joypeakv", hue="InstrResp")

plt.figure()
plt.grid()
sns.barplot(data=df, x="Congruity",y="joytotalT", hue="InstrResp")
```
